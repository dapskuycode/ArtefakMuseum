# Deep Embedded Clustering (DEC) — Improved
## Encoder: EfficientNetV2B2 Pretrained → DEC Clustering Refinement
---

**Pipeline Improved DEC:**

```
Input Gambar
      ↓
EfficientNetV2B2 (pretrained ImageNet, frozen)   ← Deep Encoder
      ↓ 1408D
Projection Head: Dense(256) → BatchNorm → Dense(64D)   ← Embedded Space
      ↓
K-Means Initialization (cluster centers)               ← DEC Init
      ↓
DEC Clustering Layer (Student-t soft assignment)       ← Clustering
      ↓
KL Divergence Loss (iterative refinement)              ← DEC Objective
```

**Justifikasi penggunaan pretrained encoder:**
> Dataset 593 gambar terlalu kecil untuk melatih autoencoder dari nol.
> Sesuai trend literatur terkini (IDEC, DeepCluster, SphericalDEC),
> encoder pretrained digunakan untuk mendapatkan representasi awal yang kuat
> sebelum DEC melakukan cluster refinement.

## Cell 4 — Bangun Arsitektur DEC
### Encoder: EfficientNetV2B2 → Projection 64D → DEC Clustering Layer

````python
# ══════════════════════════════════════════════════════════
# DEC Clustering Layer — Student-t soft assignment
# q_ij = (1 + ||z_i - u_j||^2 / alpha)^(-(alpha+1)/2)
#         / sum_j'(1 + ||z_i - u_j'||^2 / alpha)^(-(alpha+1)/2)
# ══════════════════════════════════════════════════════════
class DECLayer(layers.Layer):
    """Deep Embedded Clustering Layer.
    Mengukur kesamaan antara embedding z_i dan cluster center u_j
    menggunakan distribusi Student-t.
    """
    def __init__(self, n_clusters, alpha=1.0, **kwargs):
        super().__init__(**kwargs)
        self.n_clusters = n_clusters
        self.alpha      = alpha

    def build(self, input_shape):
        self.clusters = self.add_weight(
            shape=(self.n_clusters, input_shape[-1]),
            initializer='glorot_uniform',
            trainable=True, name='cluster_centers'
        )
        super().build(input_shape)

    def call(self, inputs):
        """Soft assignment: q_ij"""
        # ||z_i - u_j||^2
        sq_dist = tf.reduce_sum(
            tf.square(tf.expand_dims(inputs, 1) - self.clusters), axis=2
        )
        # (1 + d^2/alpha)^(-(alpha+1)/2)
        numerator = tf.pow(1.0 + sq_dist / self.alpha,
                           -(self.alpha + 1.0) / 2.0)
        q = numerator / tf.reduce_sum(numerator, axis=1, keepdims=True)
        return q

    def get_config(self):
        cfg = super().get_config()
        cfg.update({'n_clusters': self.n_clusters, 'alpha': self.alpha})
        return cfg


# ── Load fine-tuned EfficientNetV2B2 backbone ──────────────
print('Loading fine-tuned backbone dari TL_FineTuned...')
ft_model_loaded = tf.keras.models.load_model(
    CONFIG['finetuned_model_path'],
    custom_objects={},
    compile=False
)

# Ambil backbone layer (EfficientNetV2B2 di dalam model)
backbone = ft_model_loaded.get_layer('efficientnetv2-b2')
backbone.trainable = False   # Freeze backbone untuk DEC
print(f'Backbone: {backbone.name} | trainable: {backbone.trainable}')

# ── Bangun DEC Model ───────────────────────────────────────
inp     = tf.keras.Input(shape=(CONFIG['img_size'], CONFIG['img_size'], 3), name='input')
# Encoder: backbone → projection
enc_out = backbone(inp, training=False)                          # 1408D
z       = layers.Dense(256, activation='relu', name='proj1')(enc_out)
z       = layers.BatchNormalization(name='bn1')(z)
z       = layers.Dense(CONFIG['latent_dim'], name='embedding')(z)  # 64D latent
z_norm  = layers.Lambda(lambda x: tf.math.l2_normalize(x, axis=1),
                         name='l2_norm')(z)                     # unit hypersphere
# DEC clustering layer
q       = DECLayer(N_CLUSTERS, alpha=CONFIG['alpha'], name='dec_layer')(z_norm)

dec_model    = Model(inp, q, name='DEC')
enc_model    = Model(inp, z_norm, name='Encoder')  # untuk ekstraksi fitur

dec_model.summary(line_length=80)
print(f'\nLatent dim   : {CONFIG["latent_dim"]}D')
print(f'DEC clusters : {N_CLUSTERS}')
````

## Cell 7 — Definisi DEC Loss & Target Distribution
> **Target distribution P** adalah distribusi yang lebih 'tajam' dari Q.
> KL divergence KL(P||Q) mendorong cluster assignment menjadi lebih confident.

````python
def target_distribution(q):
    """Compute target distribution P dari soft assignment Q.
    p_ij = (q_ij^2 / sum_i q_ij) / sum_j (q_ij^2 / sum_i q_ij)
    Menghasilkan distribusi yang lebih tajam dari Q.
    """
    weight = (q ** 2) / q.sum(axis=0, keepdims=True)
    return (weight / weight.sum(axis=1, keepdims=True)).astype('float32')

def kl_divergence_loss(p, q):
    """KL(P||Q) = sum_ij p_ij * log(p_ij / q_ij)"""
    return tf.reduce_mean(
        tf.reduce_sum(p * tf.math.log(p / (q + 1e-9) + 1e-9), axis=1)
    )

# Optimizer untuk DEC
dec_optimizer = Adam(learning_rate=CONFIG['dec_lr'])
print('DEC Loss & Optimizer siap.')
print('Loss  : KL Divergence KL(P||Q)')
print('Optim : Adam lr={:.0e}'.format(CONFIG['dec_lr']))

# Test P distribution
p_test = target_distribution(q_init)
print(f'\nContoh Q (soft assign): {q_init[0].round(3)}')
print(f'Contoh P (target dist): {p_test[0].round(3)}')
print(f'P lebih tajam dari Q: {p_test[0].max():.3f} vs {q_init[0].max():.3f}')
````

## Cell 8 — DEC Iterative Refinement
> Inti dari DEC: iterasi mengoptimasi KL(P||Q) untuk memperketat cluster.
> Target distribution P diperbarui setiap `update_interval` iterasi.

````python
print('=== DEC ITERATIVE REFINEMENT ===')
print(f'Max iter         : {CONFIG["dec_max_iter"]}')
print(f'Update interval  : {CONFIG["dec_update_interval"]}')
print(f'Tolerance        : {CONFIG["dec_tol"]}')
print()

dec_batch_size   = CONFIG['dec_batch_size']
n_samples        = len(images_pp)
n_batches        = int(np.ceil(n_samples / dec_batch_size))
loss_history     = []
sil_history      = []
y_pred_last      = np.copy(lbl_km_dec)

t_start = time.time()

for iteration in range(CONFIG['dec_max_iter'] + 1):
    # ── Update target distribution P ────────────────────────
    if iteration % CONFIG['dec_update_interval'] == 0:
        q_all   = dec_model.predict(images_pp, batch_size=CONFIG['batch_size'], verbose=0)
        p_all   = target_distribution(q_all)
        y_pred  = np.argmax(q_all, axis=1)

        # ── Cek konvergensi ──────────────────────────────────
        delta   = np.sum(y_pred != y_pred_last).astype(float) / n_samples
        # Hitung Silhouette
        feat_cur = enc_model.predict(images_pp, batch_size=CONFIG['batch_size'], verbose=0)
        sil_cur  = silhouette_score(feat_cur, y_pred) if len(np.unique(y_pred))>1 else -1
        elapsed  = time.time() - t_start
        print(f'Iter {iteration:5d} | delta={delta:.4f} | Sil={sil_cur:.4f} | {elapsed:.0f}s')
        sil_history.append((iteration, sil_cur))
        y_pred_last = np.copy(y_pred)

        if iteration > 0 and delta < CONFIG['dec_tol']:
            print(f'\n✅ Konvergensi tercapai di iterasi {iteration} (delta={delta:.5f} < tol={CONFIG["dec_tol"]})')
            break

    # ── Training step (satu batch) ───────────────────────────
    if iteration >= CONFIG['dec_max_iter']: break
    batch_idx = (iteration % n_batches) * dec_batch_size
    x_batch   = images_pp[batch_idx : batch_idx + dec_batch_size]
    p_batch   = p_all[batch_idx : batch_idx + dec_batch_size]

    with tf.GradientTape() as tape:
        q_batch = dec_model(x_batch, training=True)
        loss    = kl_divergence_loss(
            tf.constant(p_batch, dtype=tf.float32), q_batch
        )
    grads = tape.gradient(loss, dec_model.trainable_variables)
    dec_optimizer.apply_gradients(zip(grads, dec_model.trainable_variables))
    loss_history.append(float(loss))

print('\nDEC Training selesai!')
dec_model.save(os.path.join(CONFIG['output_root'],'model','dec_model.keras'))
enc_model.save(os.path.join(CONFIG['output_root'],'model','enc_model.keras'))
````

## Cell 10 — Evaluasi Akhir: Embedding + Soft Assignment Space

````python
print('Evaluasi akhir...')
# Fitur embedding 64D (DEC latent space)
feat_dec = enc_model.predict(images_pp, batch_size=CONFIG['batch_size'], verbose=0)
# Soft assignment Q (DEC output)
q_final  = dec_model.predict(images_pp, batch_size=CONFIG['batch_size'], verbose=0)
# Label berdasarkan argmax Q
dec_labels = np.argmax(q_final, axis=1)

# ── Silhouette di berbagai ruang ──────────────────────────
sil_embed  = silhouette_score(feat_dec, dec_labels)   # DEC embedding 64D
dbi_embed  = davies_bouldin_score(feat_dec, dec_labels)
chi_embed  = calinski_harabasz_score(feat_dec, dec_labels)

sil_softq  = silhouette_score(q_final, dec_labels)    # Soft assignment space
dbi_softq  = davies_bouldin_score(q_final, dec_labels)

print(f'\n{"Ruang Evaluasi":<30} {"Silhouette":>12} {"DBI":>8}')
print('-'*53)
print(f'{"Pre-DEC (embedding 64D)":<30} {sil_init:>12.4f} {dbi_init:>8.4f}')
print(f'{"DEC embedding 64D":<30} {sil_embed:>12.4f} {dbi_embed:>8.4f}')
print(f'{"DEC soft assignment 10D":<30} {sil_softq:>12.4f} {dbi_softq:>8.4f}')
print()
print(f'Peningkatan Silhouette (64D): {sil_init:.4f} → {sil_embed:.4f}  '
      f'({((sil_embed-sil_init)/max(abs(sil_init),0.001))*100:+.1f}%)')

# ── Hungarian Accuracy ────────────────────────────────────
unique_pred = sorted(set(dec_labels))
cost = np.zeros((len(unique_pred), len(class_names)), dtype=int)
for i,cl in enumerate(unique_pred):
    for j in range(len(class_names)): cost[i,j]=np.sum((dec_labels==cl)&(gt_labels==j))
ri,ci = linear_sum_assignment(-cost)
mapping = {unique_pred[ri[k]]:ci[k] for k in range(len(ri))}
pred_m  = np.array([mapping.get(p,-1) for p in dec_labels])
acc_dec = np.mean(pred_m[pred_m!=-1]==gt_labels[pred_m!=-1])
dec_names = {cl:class_names[gi] for cl,gi in mapping.items()}

print(f'Cluster Accuracy (Hungarian): {acc_dec*100:.2f}%')
print('\nMapping DEC cluster → Kategori Museum:')
for cl,gi in sorted(mapping.items()):
    n_m=np.sum((dec_labels==cl)&(gt_labels==gi)); n_t=np.sum(dec_labels==cl)
    bar='█'*(n_m*15//max(n_t,1))
    print(f'  C{cl:02d} → {class_names[gi]:<30} {n_m:>3}/{n_t:<3}  {bar}')

np.save(os.path.join(CONFIG['output_root'],'dec_labels.npy'),  dec_labels)
np.save(os.path.join(CONFIG['output_root'],'feat_dec.npy'),    feat_dec)
np.save(os.path.join(CONFIG['output_root'],'q_final.npy'),     q_final)
````
